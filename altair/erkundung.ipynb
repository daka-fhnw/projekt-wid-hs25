# Erkundung der Daten

In diesem Jupyter Notebook wird die Erkundung der Daten vorbereitet:
Für verschiedene Zeiträume und Filterkriterien soll die Anzahl Passanten pro Standort dargestellt werden.

Einerseits werden hier die pandas-Funktionen vorbereitet für die Datenabfrage und andererseits die "spec" für die Darstellung des Diagrammes.

## Vorbereiten der Daten

### Daten aus CSV laden

```python
import pandas as pd

data = pd.read_csv('../daten/Gesamtdatensatz.csv')
data
```

### Nach Datumsbereich filtern

```python
datum_von = '2020-01-01'
datum_bis = '2025-12-31'
data = data[(data['timestamp'] >= datum_von) & (data['timestamp'] <= datum_bis)]
data
```

### Nach Stundenbereich filtern

```python
stunde_von = 0
stunde_bis = 23
stunden = pd.to_datetime(data['timestamp']).dt.hour
data = data[(stunden >= stunde_von) & (stunden <= stunde_bis)]
data
```

### Nach Temperatur filtern

```python
temperatur_von = -20
temperatur_bis = 40
data = data[(data['temperature'] >= temperatur_von) & (data['temperature'] <= temperatur_bis)]
data
```

### Gruppieren und summieren der relevanten Spalten nach Standort

```python
data = data.groupby(['location_name'], as_index=False)[[
    'pedestrians_count',
]].sum()
data
```

### Spalten umbenennen als Vorbereitung für das Diagramm

```python
data = data.rename(columns={
    'location_name': 'Standort',
    'pedestrians_count': 'Anzahl Passanten',
})
data
```

## Definition des Diagrammes

```python
import altair as alt

bar_width = 80
bar_spacing = 0.4 * bar_width
chart_height = 400

chart = alt.Chart(data).mark_bar(size=bar_width).encode(
    x=alt.X(
        'Standort:N',
        sort=['Bahnhofstrasse (Nord)', 'Bahnhofstrasse (Mitte)', 'Bahnhofstrasse (Süd)', 'Lintheschergasse'],
    ).axis(
        labelAngle=-45,
        labelOffset=bar_width / 4,
    ), 
    y=alt.Y('Anzahl Passanten:Q').axis(
        format='~s',
        tickCount=7,
    ),
    tooltip=[
        alt.Tooltip('Anzahl Passanten', format=','),
    ],
).properties(
    width={'step': bar_width + bar_spacing},
    height=chart_height
).configure_axis(
    titleFontSize=17, 
    labelFontSize=15,
)
chart
```

## Generieren der Spec für die Darstellung des Diagrammes

```python
import json

# In Dictionary umwandeln
spec = chart.to_dict()

# Dataset umbenennen in "values" für einfacheren Zugriff im Frontend
new_name = 'values'
old_name = spec['data']['name']
spec['data']['name'] = new_name
spec['datasets'][new_name] = spec['datasets'].pop(old_name)

# Spec in eine Datei schreiben
with open('erkundung.json', 'w') as f:
    json.dump(spec, f, indent=2)
```
